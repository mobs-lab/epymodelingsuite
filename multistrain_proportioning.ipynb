```python
import pandas as pd

EPIWEEK = '202614'
```

```python
multi_prop = pd.read_csv(f'../flu-forecast-epydemix/common-data/surveillance/{EPIWEEK}_multistrain_proportions.csv')
```

```python
ed = pd.read_csv(f'../flu-forecast-epydemix/common-data/surveillance/flu_prop-ed_25_{EPIWEEK}_scaled.csv')
```

```python
metro = pd.read_csv(f'../flu-forecast-epydemix/common-data/surveillance/metrocast_{EPIWEEK}_scaled.csv')
```

```python
hosp = pd.read_csv(f'../flu-forecast-epydemix/common-data/surveillance/flu_hosp_25_{EPIWEEK}_prelim.csv')
```

## Hosp

```python
us_multi_h1, us_multi_h3, us_multi_b, us_multi_total = [], [], [], []
```

```python
hosp_multi = hosp.merge(multi_prop,
                        left_on=['target_end_date', 'location_code', 'location_iso', 'epiweek'],
                        right_on=['target_end_date', 'location_code', 'location_iso', 'epiweek'],
                        how='left').drop(columns=['abbreviation']).rename(columns={'target_end_date':'date','hospitalizations':'value','location_code':'location'})
```

```python
hosp_multi['value_H1'] = hosp_multi.value * hosp_multi.prop_H1
hosp_multi['value_H3'] = hosp_multi.value * hosp_multi.prop_H3
hosp_multi['value_B'] = hosp_multi.value * hosp_multi.prop_B
```

```python
us_hosp_multi = hosp_multi[hosp_multi.location == 'US']
hosp_multi = hosp_multi[hosp_multi.location != 'US']
```

```python
for row in us_hosp_multi.itertuples():
    h1 = hosp_multi[hosp_multi.date == row.date].value_H1.sum()
    h3 = hosp_multi[hosp_multi.date == row.date].value_H3.sum()
    b = hosp_multi[hosp_multi.date == row.date].value_B.sum()
    us_multi_h1.append(h1)
    us_multi_h3.append(h3)
    us_multi_b.append(b)
```

```python
us_hosp_multi.value_H1 = us_multi_h1
us_hosp_multi.value_H3 = us_multi_h3
us_hosp_multi.value_B = us_multi_b
```

```python
hosp_multi = pd.concat([hosp_multi, us_hosp_multi])
```

```python
hosp_multi
```

```python
hosp_multi.to_csv(f'../flu-forecast-epydemix/common-data/surveillance/flu_hosp_25_{EPIWEEK}_prelim_multi.csv', index=False)
```
